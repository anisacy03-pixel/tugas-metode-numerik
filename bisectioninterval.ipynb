```python
# ==========================================
# SOAL 1: METODE BAGI DUA (BISECTION)
# f(x) = x^3 - 3, interval [1, 2]
# Rumus: c = (a + b) / 2
# ==========================================
def f1(x):
    return x**3 - 3

a, b = 1.0, 2.0
n = 4

print("SOAL 1 - Bisection: f(x) = x^3 - 3, interval [1, 2]")
print(f"{'Iter':<5}{'a':<10}{'b':<10}{'c':<12}{'f(c)':<12}")
print("-" * 49)

for i in range(1, n + 1):
    c = (a + b) / 2
    print(f"{i:<5}{a:<10.5f}{b:<10.5f}{c:<12.6f}{f1(c):<12.6f}")
    if f1(a) * f1(c) < 0:
        b = c
    else:
        a = c

print("-" * 49)
print(f"Hasil akar pendekatan: {c:.5f}\n")


# ==========================================
# SOAL 2: f(x) = x^2 - 10, interval [3, 4]
# Rumus baru: c = a + (b - a) / 2
# ==========================================
def f2(x):
    return x**2 - 10

a, b = 3.0, 4.0
n = 5

print("SOAL 2 - Rumus baru: f(x) = x^2 - 10, interval [3, 4]")
print(f"{'Iter':<5}{'a':<10}{'b':<10}{'c':<12}{'f(c)':<12}")
print("-" * 49)

for i in range(1, n + 1):
    c = a + (b - a) / 2
    print(f"{i:<5}{a:<10.5f}{b:<10.5f}{c:<12.6f}{f2(c):<12.6f}")
    if f2(a) * f2(c) < 0:
        b = c
    else:
        a = c

print("-" * 49)
print(f"Hasil akar pendekatan: {c:.5f}")
```
